# 練習 5：最近的請求次數

給一串時間 `times`，每次有一個新的請求時間 `t`。

請回傳每次在最近 3000 毫秒內有幾個請求。

範例：

```python
times = [1, 100, 3001, 3002]
```

輸出：

```python
[1, 2, 3, 3]
```

這題可以用 Queue 保留目前還在範圍內的時間。

````python
# [1, 2, 3, 4000, 7000] -> [1, 2, 3, 1, 2]

from collections import deque

def recent_counter(times):
  q = deque()
  ans = []

  for t in times:
    q.append(t)

    while q[0] < t - 3000:
      q.popleft()

    ans.append(len(q))

  return ans

print(recent_counter([1, 100, 3001, 3002]))
print(recent_counter([1, 2, 3, 4000, 7000]))
````

````python
# recursion 遞迴

# 函式自己呼叫自己，再問題太大時，先把它拆成比較小的同類型問題
````

````python
'''
倒數

countdown(3)

會印出：
3
2
1
結束
'''

def countdown(n):
  if n == 0:
    print('結束')
    return

  print(n)
  countdown(n-1)

countdown(3)
````

````python
'''
遞迴中最重要的兩部分

1. Base Case 終止條件
  - 何時停下來
  - 如果沒有 base case，程式會一直呼叫自己，會變無窮函式 -> RecursionError: maximum recursion depth exceeded
2. Recursive Case 遞迴條件
  - 把問題變小
  - 呼叫自己處理小問題
'''

def countdown(n):
  # base case
  if n == 0:
    print('結束')
    return

  # recursive case
  print(n)
  countdown(n-1)
````

````python
# Recursion vs. Stack
# 其實每次函式被呼叫時，電腦都會把它放進一個叫做 Call Stack 的地方
````

````python
'''
觀察遞迴的進入與離開

會先一路進去：
進入 3
進入 2
進入 1
進入 0

遇到 base case 後，再一層一層回來：
離開 1
離開 2
離開 3
'''

def demo(n):
  print('進入', n)

  if n == 0:
    print('遇到 base case')
    return

  demo(n - 1)

  print('離開', n)

demo(3)
````

# 範例 1：計算 1 + 2 + ... + n

問題：

```python
sum_to_n(5)
```

代表：

```python
1 + 2 + 3 + 4 + 5
```

答案是 `15`

````python
def sum_to_n(n):
  if n == 1:
    return 1

  return n + sum_to_n(n-1)

print(sum_to_n(5))
print(sum_to_n(10))
````

```python
sum_to_n(5)
= 5 + sum_to_n(4)
= 5 + 4 + sum_to_n(3)
= 5 + 4 + 3 + sum_to_n(2)
= 5 + 4 + 3 + 2 + sum_to_n(1)
= 5 + 4 + 3 + 2 + 1
= 15
```

# 範例 2：階乘 Factorial

階乘寫法：

```python
5! = 5 * 4 * 3 * 2 * 1
```

所以：

```python
factorial(5) = 120
```

````python
def factorial(n):
  if n == 1:
    return 1

  return n * factorial(n-1)

print(factorial(5))
````


```python
factorial(5)
= 5 * factorial(4)
= 5 * 4 * factorial(3)
= 5 * 4 * 3 * factorial(2)
= 5 * 4 * 3 * 2 * factorial(1)
= 5 * 4 * 3 * 2 * 1
= 120
```

# 範例 3：印出 1 到 n

這題要注意遞迴呼叫放在 print 前面還是後面，結果會不一樣。

````python
def print_1_to_n(n):
  if n == 0:
    return

  print_1_to_n(n - 1)
  print(n)

print_1_to_n(5)
````

# 範例 4：印出 n 到 1

如果把 `print(n)` 放在遞迴呼叫前面，就會變成先印再往下呼叫。

````python
def print_n_to_1(n):
  if n == 0:
    return

  print(n)
  print_n_to_1(n - 1)

print_n_to_1(5)
````

# 範例 5：字串反轉

問題：

```python
reverse('hello')
```

答案：

```python
'olleh'
```

````python
def reverse(s):
  if len(s) == 0:
    return ''

  return reverse(s[1:]) + s[0]

print(reverse('hello'))
print(reverse('python'))
````

```python
= reverse('') + 'o' + 'l' + 'l' + 'e' + 'h'

```

# 範例 6：二分搜尋 Binary Search 遞迴版
前提：資料必須已經排序好

````python
# 一般版

def binary_search(arr, target):
  low, high = 0, len(arr) - 1

  while low <= high:
    mid = (low + high) // 2

    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      low = mid + 1
    else:
      high = mid - 1

  return -1

binary_search([1, 3, 5, 7, 9], 7)
````

````python
# 遞迴版

def binary_search(arr, target, low, high):
  # base case
  if low > high:
    return -1

  # recursive case
  mid = (low + high) // 2

  if arr[mid] == target:
    return mid

  if arr[mid] > target:
    return binary_search(arr, target, low, mid - 1)
  else:
    return binary_search(arr, target, mid + 1, high)


arr = [1, 3, 5, 7, 9, 11, 13]
print(binary_search(arr, 7, 0, len(arr) - 1))
print(binary_search(arr, 8, 0, len(arr) - 1))
````

````python
def search(arr, target):
    if len(arr) == 0:
        return -1

    mid = len(arr) // 2

    if target == arr[mid]:
        return mid
    elif target < arr[mid]:
        return search(arr[:mid], target)
    else:
        result = search(arr[mid+1:], target)

        if result == -1:
            return -1
        return mid + 1 + result
````

# 河內塔 Tower of Hanoi

河內塔是經典遞迴題。

規則：

1. 有三根柱子 A、B、C
2. 一次只能移動一個盤子
3. 大盤子不能放在小盤子上面
4. 目標是把所有盤子從 A 移到 C

````python
def hanoi(n, start, helper, end):
  # base case
  if n == 1:
    print('把盤子從', start, '移到', end)
    return

  # recursive case
  hanoi(n - 1, start, end, helper)
  print('把盤子從', start, '移到', end)
  hanoi(n - 1, helper, start, end)

hanoi(3, 'A', 'B', 'C')
````
