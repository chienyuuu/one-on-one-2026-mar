```python
'''
檔案

1. open(檔案名稱, 模式, 編碼)：只有名稱是必填、結束需要使用「檔案物件.close()」
  - f = open(....)
  - f.close()
2. 模式
  - r: 讀取
  - w: 寫入，若沒有此檔案，會自己建立一個，若有的話就覆蓋原檔案內容
  - a: 附加寫入，若沒有此檔案，會自己建立一個，若有的話就寫在原檔案後面
  - r+: 可讀寫(只能使用在文件存在情況)
  - w+: 可讀寫
  - a+: 可讀寫
3. with open(檔案名稱, 模式) as 檔案物件 (with ... as：語法結束會自動關閉檔案)
4. read()：讀取指定長度字元，沒有指定就讀全部
5. readlines()：讀取全部，回傳成「串列」，一列是一個元素
6. readline()：讀取此列「長度」的內容。有參數就讀該長度，沒有就讀整列
'''
```

```python
# 檔案和目錄管理

# os.path 模組：檢查檔案、路徑是否存在，檢查檔案大小

# isfile()：檢查「檔案」是否存在
# isdir()：檢查「路徑」是否存在
# exists()：同時檢查「檔案/資料夾」是否存在

import os

print(os.path.isfile('test.txt'))
print(os.path.isfile('test1.txt'))
print()

print(os.path.exists('test.txt'))
print(os.path.exists('test1.txt'))
print()

print(os.path.isdir('class8'))
print(os.path.isdir('class9'))
print()

print(os.path.exists('class8'))
print(os.path.exists('class9'))
print()

print(os.path.isfile('class8/class8.txt'))
print(os.path.exists('class8/class8.txt'))
```

```python
# remove()：刪除檔案，通常會配合 exists()，要確定存在才可以刪除

import os

file = 'test.txt'
if os.path.exists(file):
  os.remove(file)
  print('ok')
else:
  print("file doesn't exist")
```

```python
# mkdir()：建立資料夾

import os

folder = 'test_dir'
if not os.path.isdir(folder):
  os.mkdir(folder)
  print('ok')
else:
  print('路徑已存在')
```

```python
# rmdir()：刪除資料夾

import os

folder = 'test_dir'
if os.path.isdir(folder):
  os.rmdir(folder)
  print('ok')
else:
  print('路徑不存在')
```

```python
# 例外處理

# 變數不存在、資料型態不符合、檔案不存在 -> 不想中斷程式執行

'''
try:
  測試的程式碼
except 例外情況:
  處理指定的例外情況
except:
  處理其他所有例外情況
else:
  沒有發生例外情況時執行
finally:
  一定會執行（通常是刪除物件/關閉檔案）
'''

# 至少要有一個 except
# else, finally 可有可無
# except 範圍要由小而大
```

```python
n = 10

try:
  print(n)
except:
  print('變數 n 不存在')
else:
  print('變數 n 存在')
finally:
  print('不管怎樣都執行')
```

```python
# 兩個整數求和

try:
  a = int(input('a = '))
  b = int(input('b = '))
  res = a + b
  print('res =', res)
except:
  print('輸入格式有問題')
```

```python
'''
常用錯誤表

1. IOError：輸入/輸出錯誤
2. NameError：變數名稱未宣告錯誤
3. ZeroDivisionError：除數為 0 錯誤
4. ValueError：輸入非數值錯誤
'''
```

```python
# 輸入兩正整數，求相除的餘數

try:
  a = int(input('a = '))
  b = int(input('b = '))
  res = a % b
  print('res =', res)
except ValueError:
  print('非數值錯誤')
except Exception as e:
  print('發生', e, '錯誤（包含除數是 0）')
finally:
  print('不管怎樣都執行')
```

```python
'''
Stack 常見操作：

push：放一個資料進去 stack
pop ：拿出最上面的資料
top ：查看最上面的資料，但不拿出來
is_empty：檢查 stack 是否為空
'''

stack = []

stack.append(10) # push
stack.append(20) # push
stack.append(30) # push

print(stack)

x = stack.pop() # pop
print(x)
print(stack)

print(stack[-1]) # top

print(len(stack) == 0) # is_empty
```

```python
# 手刻 stack

def push(stack, value):
  stack.append(value)

def pop(stack):
  if len(stack) == 0:
    return None
  return stack.pop()

def top(stack):
  if len(stack) == 0:
    return None
  return stack[-1]

def is_empty(stack):
  return len(stack) == 0

s = []

push(s, 5)
push(s, 10)
push(s, 15)

print(s)
print(top(s))
print(pop(s))
print(pop(s))
print(s)
print(is_empty(s))
```
