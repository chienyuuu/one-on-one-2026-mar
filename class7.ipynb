```python
'''
1. 排序
  - 泡沫排序法 bubble sort: 做法是重複的兩兩比較並交換位置，若是從小排到大，每一輪結束會有一值排至最右邊，O(n²)
  - 選擇排序法 selection sort: 每一輪從「還沒排好的數字中」找最小的，放到最前面，若是從小排到大，每一輪結束會有一值排至最左邊，O(n²)
2. 搜尋
  - 循序搜尋法 linear seach: 資料不用排序，從第一個開始逐一比對，找到就 break，沒有找到就繼續比對下一個，直到串列全部搜尋完，O(n)
  - 二分搜尋法 binary search: 資料「要」排序，先找出中間元素，根據比大小移動上下界，O(log n)
3. 時間複雜度: 用來表示「演算法需要花多少時間」的方式
  - O(1)    → 固定時間（最快）
  - O(log n)→ 很快（資料越多，增加很慢）
  - O(n)    → 一個一個看
  - O(n²)   → 很慢（資料多會爆掉）

'''
```

```python
# 物件 object

'''
可以想像成一個人，每個人都有自己的屬性，例如：年齡、性別、身高

如果我們要使用一個人的物件，我們可以像這樣使用它：

print(People.age)
print(People.height)
'''
```

```python
# 類別 class

class Person:
  def __init__(self, age, height): # 初始化函式
    self.age = age
    self.height = height

  def birthday(self): # 其他自訂函式
    self.age += 1

amy = Person(20, 160)
print(amy.age)
print(amy.height)

amy.birthday()
print(amy.age)
```

```python
'''
(1) class 類別名稱 : 類別關鍵字
類通常自訂類別使用大寫作為開頭，例如：Person

(2) init(self): 初始化函式
初始化函式用來定義該類別的初始屬性，例如：age, height
其中的 self 代表當下的物件本身，所以可以看成“自己的age“

(3)自定義函式：
使用自定義函式時，可以將物件本身進行改動，例如：改變屬性等
若要使用物件屬性，需要將“self”當作參數，使用self.屬性即可讀取
'''
```

```python
class Car:
  def __init__(self, brand, price):
    self.brand = brand
    self.price = price

  def discount(self, discount):
    self.price *= discount

  def __str__(self):
    return f"{self.brand} - {self.price}"

car1 = Car('toyota', 100)
car2 = Car('benz', 1000)

print(car1)
print(car2)

car2.discount(0.8)
print(car2.price)
```

```python
# 檔案

# open(檔案名稱, 模式, 編碼)：：只有檔案名稱必填

'''
模式

1. r：讀取
2. w：寫入，若沒有此檔案，會自己建立一個，若有的話就覆蓋原檔案內容
3. a：附加寫入，若沒有此檔案，會自己建立一個，若有的話就寫在原檔案後面
4. r+：可讀寫
5. w+：可讀寫
6. a+：可讀寫

'r'、'r+' ：只讀或讀寫，並不創建檔案
'w'、'w+'、'a'、'a+'，若檔案不存在都會新建檔案
'w'、'w+' 只寫或讀寫，兩者都等同新建一個檔案，故會把原內容清除
'w+' 與 'r+'的區別：'r+'只能使用在文件存在情況，不存在報Error
'''
```

```python
f = open('test.txt', 'r') # 建立物件處理檔案

for line in f:
    print(line, end='')

f.close() # 關閉檔案
```

```python
# 用 open 就要用 close
# with ... as：語法結束會自動關閉檔案

# with open(檔案, 模式) as 檔案物件:
with open('test.txt', 'r') as f:
  for line in f:
    print(line, end='')
```

```python
# read()：讀取指定長度字元，沒有指定就讀全部

f = open('test.txt', 'r')

print(f.read(3))
print(f.read())

f.close()
```

```python
# readlines()：讀取全部，回傳成串列，一列是一個元素
with open('test.txt', 'r') as f:
  content = f.readlines()
  print(content)
  print(type(content))
```

```python
# readline(長度)：讀取此列「長度」的內容。有參數就讀該長度，沒有就讀整列
f = open('test.txt', 'r')

print(f.readline())
print(f.readline())
print(f.readline(1))

f.close()
```
