```python
# 排序

'''
泡沫排序法 bubble sort

最簡單、常見，做法是重複的兩兩比較並交換位置

由小排到大
1. 相鄰元素比較，左大於右就交換
2. 每一對都做一樣的事，從第一對到結尾那對。結束後最大的會在最右邊
3. 下一次再繼續從頭到尾做一樣的事，除了最後一個
4. 持續對越來越少的元素去比較並交換，直到沒有任何一個數字需要比較
'''
```

```python
arr = [5, 3, 8, 4, 2]
n = len(arr)

for i in range(n):
  for j in range(n - i - 1):
    if arr[j] > arr[j+1]:
      arr[j], arr[j+1] = arr[j+1], arr[j]

print(arr)
```

```python
'''
練習：寫出從大排到小的泡沫排序法

[2, 3, 5, 7, 1]
'''
```

```python
'''
選擇排序法 Selection Sort

每一輪從「還沒排好的數字中」找最小的，放在最前面

由小排到大
1. 從第一個位置開始，假設它是最小值
2. 往後找，如果有更小的就記錄下來
3. 找完後，把最小的數字跟目前位置交換
4. 接著換下一個位置，重複一樣的動作
5. 每一輪都會把一個最小的數字放到正確位置
'''
```

```python
arr = [5, 3, 8, 4, 2]
n = len(arr)

for i in range(n):
  min_index = i # 假設目前是最小的

  for j in range(i+1, n):
    if arr[j] < arr[min_index]:
      min_index = j

  # 把最小的交換到前面
  arr[i], arr[min_index] = arr[min_index], arr[i]

print(arr)
```

```python
'''
二分搜尋法 Binary Search

需要先排序！！！

1. 資料排序（通常從小排到大）
2. 用中間元素切成 小元素部分、大元素部分
3. 欲找元素和中間元素比較
4. 找到就結束搜尋
5. 欲找元素 > 中間元素 -> 從大元素部分繼續找
6. 欲找元素 < 中間元素 -> 從小元素部分繼續找
'''
```

```python
def binary_search(arr, target):
  low, high = 0, len(arr) - 1
  while low <= high:
    mid = (low + high) // 2

    if arr[mid] == target:
      return mid
    elif arr[mid] < target:
      low = mid + 1
    else:
      high = mid - 1
  return -1

binary_search([1, 3, 5, 7, 9], 9)
```

```python
'''
串列元素有 n 個，搜尋次數是 m 次

n 為 2 的指數 -> 2^(m-1) = n
n 不為 2 的指數 -> 2^m > n


若有 8 筆資料 (2^3) -> 最多 4 次可找到
若有 20 筆資料，最多 5 次可找到
'''
```

```python
# 時間複雜度 time complexity

'''
表示「演算法需要花多少時間」的方式

重點不是實際秒數，而是當資料變多時，速度會變慢多少

常用表示法：Big O（大 O 表示法）
'''
```

```python
# 常見的時間複雜度

'''
O(1)    → 固定時間（最快）
O(log n)→ 很快（資料越多，增加很慢）
O(n)    → 一個一個看
O(n²)   → 很慢（資料多會爆掉）
'''
```

```python
# 範例

'''
假設有 1000 (n) 筆資料：

O(n)    → 最多要找 1000 次
O(log n)→ 大約只要找 10 次
O(n²)   → 可能要 1,000,000 次
'''
```

```python
'''
泡沫排序（Bubble Sort）：
→ O(n²)

選擇排序（Selection Sort）：
→ O(n²)

兩個都屬於「慢的排序」（但好理解）
'''
```

```python
'''
循序搜尋（Sequential/Linear Search）：
→ O(n)
→ 一個一個找

二分搜尋（Binary Search）：
→ O(log n)
→ 每次砍一半（但一定要「已排序」）
'''
```
