```python
# 模組 module：擴充功能

# 亂數模組 random

# 使用 import 匯入
import random

# 模組中有許多函式，使用方式：模組.函式(參數)
random.randint(參數)

# 若只使用模組中其中一個函式
from random import randint
randint(參數)

# 若有兩個模組有一樣的函式，可以幫它命名（簡短又避免重複）
import random as r
r.randint(參數)

# 模組中的函式也可以命名
from random import randint as rt
rt(參數)
```

```python
# random.randint(起始值, 終止值)：指定範圍產生整數亂數
# 注意：這裡包含起始值也包含終止值

import random

for i in range(5):
  print(random.randint(1, 10))
```

```python
# random.randrange(起始值, 終止值, 遞增值)
# 遞增值可有可無，沒有的話預設 1
# 注意：這裡包含起始值但「不」包含終止值

import random

for i in range(5):
  print(random.randrange(0, 10, 2)) # 0, 2, 4, 6, 8
```

```python
# random.random()：產生 0 ~ 1 的浮點數亂數

import random

print(random.random())
```

```python
# random.uniform(起始值, 終止值)：指定範圍的浮點數亂數
# 注意：這裡包含起始值也包含終止值

import random

print(random.uniform(3, 10))
```

```python
'''
練習：用亂數隨機擲 3 次骰子，印出三次的值，以及總和

骰子點數是 1 ~ 6
'''
```

```python
# random.choice(字串或串列)：隨機取得其中一個字元或元素

# 字串
import random

for i in range(5):
  print(random.choice('abcdefg'), end=',')
print()

# 串列
import random

for i in range(5):
  print(random.choice([1, 2, 3, 4, 5, 6]), end=',')
```

```python
# random.sample(字串或串列, 數量)：與 chioce 類似，可取得多個

import random

print(random.sample('abcdefg', 3))
print(random.sample([1, 2, 3, 4, 5, 6], 4))

# 注意：數量不可大於元素長度，也不可是負數
print(random.sample([1, 2, 3, 4, 5, 6], 7)) # error!
```

```python
# 時間模組 time

# 取得系統時間、計算時間差
# python 時間以 tick 為單位，百萬分之一（微秒）
# 時間數值：1970 年 1 月 1 日 零時
# 精確到小數點後六位的浮點數
```

```python
# time.time()：取得目前「時間數值」

import time

print(time.time()) # 從 1970 年 1 月 1 日 零時 到現在經過幾秒
```

```python
# time.localtime(時間數值)：以時間數值計算時間元組（時間數值可有可無，沒有就是現在）

import time

print(time.localtime())
print(time.localtime(time.time()))

# 取得單一項目
time1 = time.localtime()
print(time1.tm_mon)
print(time1[1])
```

```python
# time.ctime(時間數值)：與 localtime 類似，但得到的是字串（時間數值可有可無，沒有就是現在）

import time

print(time.ctime())
print(time.ctime(time.time()))
```
