````python
'''
1. isfile()：檢查「檔案」是否存在
2. isdir()：檢查「路徑」是否存在
3. exists()：同時檢查「檔案/路徑」是否存在
4. remove()：刪除檔案，通常會配合 exists()，要確定存在才可以刪除
5. mkdir()：建立資料夾
6. rmdir()：刪除資料夾
7. 例外處理：當不想中斷程式執行
  try:
    測試的程式碼
  except 例外情況:
    處理指定的例外情況
  except:
    處理其他所有例外情況
  (else):
    沒有發生例外情況時執行
  (finally):
    一定會執行（通常是刪除物件/關閉檔案）
8. 常用錯誤表
  - IOError：輸入/輸出錯誤
  - NameError：變數名稱未宣告錯誤
  - ZeroDivisionError：除數為 0 錯誤
  - ValueError：輸入非數值錯誤
9. stack 堆疊：last in first out 後進先出
  - push：放一個資料進去 stack
  - pop ：拿出最上面的資料
  - top ：查看最上面的資料，但不拿出來
  - is_empty：檢查 stack 是否為空
'''
````

````python
# 手刻 stack

def push(stack, value):
  stack.append(value)

def pop(stack):
  if len(stack) == 0:
    return None
  return stack.pop()

def top(stack):
  if len(stack) == 0:
    return None
  return stack[-1]

def is_empty(stack):
  return len(stack) == 0

s = []

push(s, 5)
push(s, 10)
push(s, 15)

print(s)
print(top(s))
print(pop(s))
print(pop(s))
print(s)
print(is_empty(s))
````

````python
# 用 class 手刻 stack

class Stack:
  def __init__(self):
    self.data = []

  def push(self, value):
    self.data.append(value)

  def pop(self):
    if self.is_empty():
      return None
    return self.data.pop()

  def top(self):
    if self.is_empty():
      return None
    return self.data[-1]

  def is_empty(self):
    return len(self.data) == 0

  def size(self):
    return len(self.data)

s = Stack()

s.push(100)
s.push(200)
s.push(300)

print(s.top())
print(s.pop())
print(s.pop())
print(s.size())
````

# 練習 1：括號配對

請寫一個函式 `is_valid_parentheses(s)`，判斷括號是否正確配對。

只會出現：`()[]{}`

範例：

```python
is_valid_parentheses('()[]{}') # True
is_valid_parentheses('([{}])') # True
is_valid_parentheses('(]')     # False
is_valid_parentheses('([)]')   # False
```

````python
def is_valid_parentheses(s):
  stack = []
  pair = {
      ')': '(',
      ']': '[',
      '}': '{'
  }

  for ch in s:
    if ch in '([{':
      stack.append(ch)
    else:
      if len(stack) == 0:
        return False

      top = stack.pop()

      if top != pair[ch]:
        return False
  return len(stack) == 0

print(is_valid_parentheses('()[]{}'))
print(is_valid_parentheses('([{}])'))
print(is_valid_parentheses('(]'))
print(is_valid_parentheses('([)]'))
print(is_valid_parentheses('('))
````

# 練習 2：反轉字串

請寫一個函式 `reverse_string(s)`，用 Stack 把字串反轉。

範例：

```python
reverse_string('hello')
```

輸出：

```python
'olleh'
```

````python
def reverse_string(s):
  stack = []

  for ch in s:
    stack.append(ch) # push

  ans = ''

  while len(stack) > 0:
    ans += stack.pop() # pop

  return ans

print(reverse_string('hello'))
print(reverse_string('python'))
````

# 練習 3：移除相鄰重複字元

給一個字串，如果有兩個相鄰且相同的字元，就把它們消掉。

用 Stack 解。

範例：

```python
remove_duplicates('abbaca')
```

過程：

```python
abbaca -> aaca -> ca
```

答案：

```python
'ca'
```

````python
# azxxzyx -> azzyx -> ayx
# abbaca -> aaca -> ca

def remove_duplicates(s):
  stack = []

  for ch in s:
    if len(stack) > 0 and stack[-1] == ch:
      stack.pop()
    else:
      stack.append(ch)

  return ''.join(stack)


print(remove_duplicates('abbaca'))
print(remove_duplicates('azxxzyx'))
````

````python
# queue 佇列

# 先進先出 First In First Out
# 先排隊的人，會先被服務
````

````python
'''
Queue 常見操作：

enqueue：把資料放到隊伍最後面
dequeue：把資料從隊伍最前面拿出來
front：查看最前面的資料
is_empty：檢查 queue 是否為空
'''

from collections import deque

q = deque()

q.append(10) # enqueue
q.append(20)
q.append(30)

print(q)

x = q.popleft() # dequeue
print(x)
print(q)

print(q[0]) # front

print(len(q) == 0) # is_empty
````

````python
'''
為什麼是 popleft() 而不是 pop(0)

Python list 的 pop(0) 會把第一個元素拿掉
但是拿掉後，後面的每個元素都要往前移動，所以效率比較差

資料少的時候沒差，資料多的時候建議使用 collections.deque
'''
````

````python
# 手刻 queue

from collections import deque

def enqueue(queue, val):
  queue.append(val)

def dequeue(queue):
  if len(queue) == 0:
    return None
  return queue.popleft()

def front(queue):
  if len(queue) == 0:
    return None
  return queue[0]

def is_empty(queue):
  return len(queue) == 0


q = deque()

enqueue(q, 5)
enqueue(q, 8)
enqueue(q, 12)

print(q)
print(front(q))
print(dequeue(q))
print(dequeue(q))
print(q)
````

````python
# 用 Class 手刻 Queue

from collections import deque

class Queue:
  def __init__(self):
    self.data = deque()

  def enqueue(self, value):
    self.data.append(value)

  def dequeue(self):
    if self.is_empty():
      return None
    return self.data.popleft()

  def front(self):
    if self.is_empty():
      return None
    return self.data[0]

  def is_empty(self):
    return len(self.data) == 0


q = Queue()

q.enqueue('A')
q.enqueue('B')
q.enqueue('C')

print(q.front())
print(q.dequeue())
print(q.dequeue())
````

# 練習 4：模擬排隊

有一群人依序排隊，每次服務最前面的人。

請寫一個函式 `serve_people(names)`，回傳服務順序。

範例：

```python
serve_people(['Amy', 'Bob', 'Cindy'])
```

輸出：

```python
['Amy', 'Bob', 'Cindy']
```

````python
from collections import deque

def serve_people(names):
  q = deque()

  for name in names:
    q.append(name)

  order = []

  while len(q) > 0:
    person = q.popleft()
    order.append(person)

  return order

print(serve_people(['Amy', 'Bob', 'Cindy']))
````
